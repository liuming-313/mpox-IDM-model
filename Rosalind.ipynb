```python
## The probability that two randomly selected mating organisms will produce an individual possessing a dominant allele (and thus displaying the dominant phenotype). Assume that any two organisms can mate.

# Given: Three positive integers k, m, and n, representing a population containing k+m+n organisms: k individuals are homozygous dominant for a factor, m are heterozygous, and n are homozygous recessive.

# Return: The probability that two randomly selected mating organisms will produce an individual possessing a dominant allele (and thus displaying the dominant phenotype). Assume that any two organisms can mate.

def probability_dominant(k, m, n):
    total = k + m + n
    #probability of selecting two homozygous dominant
    p1 = (k / total) * ((k - 1) / (total - 1))
    #probability of selecting one homozygous dominant and one heterozygous
    p2 = (k / total) * (m / (total - 1)) + (m / total) * (k / (total - 1))
    #probability of selecting one homozygous dominant and one homozygous recessive
    p3 = (k / total) * (n / (total - 1)) + (n / total) * (k / (total - 1))
    #probability of selecting two heterozygous
    p4 = (m / total) * ((m - 1) / (total - 1)) * 0.75
    #probability of selecting one heterozygous and one homozygous recessive
    p5 = (m / total) * (n / (total - 1)) * 0.5 + (n / total) * (m / (total - 1)) * 0.5
    #
    print(p1, p2, p3, p4, p5)
    return p1 + p2 + p3 + p4 + p5

#test the function with the given values
probality = probability_dominant(2,2,2)
print(probality)
```

```python
#Finding a most likely common ancestor
# Given: A collection of at most 10 DNA strings of equal length (at most
# 1 kbp) in FASTA format. Return: A consensus string and profile matrix for the collection. (If several possible consensus strings exist, then you may return any one of them.)
# ATGCAACT
# A: 5 1 0 0 5 5 0 0
# C: 0 0 1 4 2 0 6 1
# G: 1 1 6 3 0 1 0 0
# T: 1 5 0 0 0 1 1 6
def consensus_and_profile(dna_strings):
    consensus = ''
    profile = {'A': [], 'C': [], 'G': [], 'T': []}
    for i in range(len(dna_strings[0])):
        count = {'A': 0, 'C': 0, 'G': 0, 'T': 0}
        for dna in dna_strings:
            count[dna[i]] += 1
        consensus += max(count, key=count.get)
        for base in profile:
            profile[base].append(count[base])
    return consensus, profile

#test the function with the given dna strings
#read the dna strings from the fasta file
fasta_file= open('C:/Users/Administrator/Downloads/rosalind_cons.txt', 'r')
dna_strings = []
current_dna = ''
for line in fasta_file:
    if line.startswith('>'):
        if current_dna:
            dna_strings.append(current_dna)
            current_dna = ''
    else:
        current_dna += line.strip()
if current_dna:
    dna_strings.append(current_dna)
consensus, profile = consensus_and_profile(dna_strings)
print(consensus)
print(profile)
```
